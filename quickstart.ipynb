# adehub quickstart

Load a dataset from [adehub.cloud](https://adehub.cloud) into pandas, run SQL over it, and push a result back.

You need an access token from **Settings → Access Tokens**. In Colab, store it as a secret named `ADEHUB_TOKEN` (key icon in the left sidebar) and grant this notebook access; `adehub.login()` then needs no input.

```python
%pip install -q "adehub[data]==0.1.0" pandas
```

```python
import adehub

adehub.login()
```

```python
REPO = "owner/slug"  # @param {type:"string"}
```

```python
df = adehub.load_dataset(REPO, format="pandas")
df.head()
```

```python
# The dataset is exposed to SQL as table `t`.
adehub.query(REPO, "select count(*) as rows from t", format="pandas")
```

## Push a result back (optional)

This publishes the first 100 rows as a new **private** dataset under your own namespace, in the same category as the source. Re-running pushes a new version.

```python
owner, slug = REPO.split("/", 1)
source = adehub.ADEClient().resolve_repo(owner, slug) or {}
me = adehub.whoami()

result = adehub.push(
    df.head(100),
    repo=f"{me['namespace']}/{slug}-derived",
    message="derived in the adehub quickstart",
    category=source.get("category", "other"),
)
print(result.url)
```
