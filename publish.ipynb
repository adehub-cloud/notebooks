# Publish a dataset to adehub

Build a DataFrame, publish it to [adehub.cloud](https://adehub.cloud) with one call, and open it in the browser.

You need an access token with **write** permission from **Settings → Access Tokens**. In Colab, store it as a secret named `ADEHUB_TOKEN` and grant this notebook access.

```python
import adehub

adehub.login()
```

```python
REPO = "my-first-dataset"  # @param {type:"string"}  (owner/slug, or just a slug for your own namespace)
CATEGORY = "other"  # @param ["finance", "maritime", "telecom", "geo", "image", "audio", "text", "health", "retail", "energy", "government", "cad", "other"]
VISIBILITY = "private"  # @param ["private", "public"]
```

```python
import pandas as pd

# Replace this with your own data: anything pandas or pyarrow can hold, or a path to a CSV/Parquet file.
df = pd.DataFrame({
    "city": ["Jakarta", "Surabaya", "Bandung"],
    "population_2025": [10_600_000, 2_900_000, 2_500_000],
})
df
```

```python
result = adehub.push(df, repo=REPO, message="initial", category=CATEGORY, visibility=VISIBILITY)
print(f"v{result.version}: {result.url}")
```
